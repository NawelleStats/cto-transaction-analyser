### Ajout de données manquantes à la main
Afin de les récupérer, j'ai du :
- identifier le mois avec des données manquantes,
- récupérer les avis d'opérés sur mon compte,
- et reporter à la main les données dans liste qui contient des dictionnaires associés à chaque transaction.

Dans mon cas, il n'y avait pas beaucoup d'opération à reporter donc j'ai pu le faire manuellement (comme à l'ancienne). Dans le cas d'un gros volume à traiter, un LLM peu traiter tous les avis d'opérés pour construire les données à ajouter ci-dessous.

```python
donnees_manquantes = [
    {
        "Date opération": "04/09/2025",
        "Date valeur": "04/09/2025",
        "Opération": "ACHAT COMPTANT",
        "Valeur": "LVMH MOET VUITTON",
        "Code ISIN": "FR0000121014",
        "Montant": -503.99,
        "Quantité": 1,
        "Cours": 500.00
    },
    {
        "Date opération": "15/09/2025",
        "Date valeur": "15/09/2025",
        "Opération": "VENTE COMPTANT",
        "Valeur": "KERING",
        "Code ISIN": "FR0000121485",
        "Montant": 482.01,
        "Quantité": 2,
        "Cours": 242
    }
]
```

```python
# ajout des données manquantes, tri par date, valeur et opération, puis sauvegarde du fichier CSV corrigé
df = pd.concat([df, pd.DataFrame(donnees_manquantes)], ignore_index=True)
df = df.sort_values(by=["Date opération", "Valeur", "Opération"], ascending=[True, True, False])
```

```python
# enregistrement
df.to_csv(CSV_PATH + FILE_NAME, sep=";") # en reprenant le chemin original on écrase le fichier CSV avec les corrections apportées
```
